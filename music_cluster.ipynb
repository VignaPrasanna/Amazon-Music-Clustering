**FEATURE SCALING**

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X = scaler.fit_transform(df1)
```

```python
X_scaled = pd.DataFrame(X, columns = df1.columns)
```

```python
X_scaled
```

```python
from sklearn.cluster import KMeans
inertia = []
for k in range(2, 11):
    kmeans = KMeans(n_clusters = k, max_iter = 300, n_init = 10, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
K = range(2,11)
plt.plot(K, inertia, marker = 'o')
plt.show()
```

**USING KNEELOCATOR TO FIND THE BEST K VALUE**

```python
from kneed import KneeLocator

# Automatically find the elbow point
kl = KneeLocator(
    list(K), inertia, curve="convex", direction="decreasing"
)
optimal_k = kl.elbow
print(f"The optimal number of clusters is: {optimal_k}")
optimal_inertia = inertia[list(K).index(optimal_k)]
print(f"Optimal inertia: {optimal_inertia}")
```

```python
optimal_inertia = inertia[list(K).index(optimal_k)]

plt.figure(figsize=(8, 8))
plt.plot(K, inertia, marker="o", linewidth=2, label="Inertia Curve")

plt.scatter(optimal_k,optimal_inertia,color="red", s=50, zorder = 3)
plt.axvline(x = optimal_k, color = 'gray', linestyle ='--')
plt.axhline(y = optimal_inertia, color = 'gray', linestyle ='--')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.xticks(K)
plt.show()
```

**K-Means Clustering**

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_scaled)
```

```python
df_og = pd.read_csv(r"D:\DS\MiniProjects\Amazon Music Clustering\single_genre_artists.csv")
```

```python
df_og["Cluster_Label"] = cluster_labels
```

**SILHOUETTE SCORE**

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X_scaled, cluster_labels)

print(f"Silhouette Score: {score:.4f}")
```

**DAVID_BOULDIN INDEX**

```python
from sklearn.metrics import  davies_bouldin_score
db_score = davies_bouldin_score(X_scaled, cluster_labels)
print(db_score)
```

**PCA ( Principal Component Analysis)**

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2, random_state=42)
pca.fit_transform(X_scaled)
pca.explained_variance_ratio_

#X_pca = pca.fit_transform(X_scaled)
#df_pca = pd.DataFrame(X_pca, columns = ["PC1", "PC2"])
```

```python
X_pca = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
pca_df["Cluster"] = df_og["Cluster_Label"]

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="Cluster",
    palette="tab10",
    s=40,
    alpha=0.7,
)

plt.title("PCA Cluster Visualization", fontsize=15)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend(title="Cluster")
plt.grid(alpha=0.3)
plt.show()
```

```python
scaled_plot_df = X_scaled.copy()
scaled_plot_df["Cluster_Label"] = df_og["Cluster_Label"]
scaled_cluster_means = scaled_plot_df.groupby("Cluster_Label").mean()

ax = scaled_cluster_means.T.plot(kind="bar", figsize=(14, 7), width=0.8)
plt.title("Average Feature Values per Cluster")
plt.xlabel("Audio Features")
plt.ylabel("Mean")
plt.xticks(rotation=45, ha="right")
plt.axhline(0, color="black", linewidth=0.8, linestyle="-")
plt.legend(title="Cluster Label", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.heatmap(
    scaled_cluster_means,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    cbar=True,
)
plt.title("Heatmap comparing features across clusters")
plt.xlabel("Audio Features")
plt.ylabel("Cluster Label")
plt.tight_layout()
plt.show()
```
